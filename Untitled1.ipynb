```python
# ==============================================================================
# STEP 1: IMPORT REQUIRED LIBRARIES
# ==============================================================================
# NLTK (Natural Language Toolkit) is standard python library for working with human language data.
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer

# Pandas handles tabular data structures (DataFrames), similar to an in-memory Excel sheet.
import pandas as pd

# NumPy provides support for high-performance mathematical functions and arrays.
import numpy as np

# Plotly Express creates interactive data visualizations (charts, bar graphs, box plots).
import plotly.express as px

# Download VADER's pre-trained sentiment dictionary/lexicon from NLTK.
nltk.download('vader_lexicon')


# ==============================================================================
# STEP 2: LOAD AND PREPARE DATA
# ==============================================================================
# File path for the uploaded CSV dataset in Google Colab.
file_path = '/content/ownwell_reviews.csv'

# Load CSV file into a Pandas DataFrame.
df = pd.read_csv(file_path)

# Extract numeric digits from the 'rating' column (e.g., transforms string "5 stars" into numeric float 5.0).
# '\d+' is a Regular Expression (RegEx) matching one or more numerical digits.
df['stars_numeric'] = df['rating'].str.extract('(\d+)').astype(float)

# Handle Missing Data: Replace null or empty values in 'review_text' with empty strings to prevent errors.
df['review_text'] = df['review_text'].fillna('')


# ==============================================================================
# STEP 3: RUN VADER SENTIMENT ANALYSIS (NLP)
# ==============================================================================
# Instantiate the VADER Sentiment Analyzer model.
sia = SentimentIntensityAnalyzer()

# Apply VADER to every review text to compute raw sentiment scores.
# The result is a dictionary containing: {'neg': float, 'neu': float, 'pos': float, 'compound': float}
df['sentiment_scores'] = df['review_text'].apply(
    lambda text: sia.polarity_scores(text)
)

# Extract individual score components into dedicated columns for granular analysis.
df['compound'] = df['sentiment_scores'].apply(lambda x: x['compound'])
df['pos_score'] = df['sentiment_scores'].apply(lambda x: x['pos'])
df['neu_score'] = df['sentiment_scores'].apply(lambda x: x['neu'])
df['neg_score'] = df['sentiment_scores'].apply(lambda x: x['neg'])


# Helper Function: Classify sentiment based on standard NLTK threshold benchmarks.
def classify_sentiment(compound_score):
  # Scores >= 0.05 are classified as Positive.
  if compound_score >= 0.05:
    return 'Positive'
  # Scores <= -0.05 are classified as Negative.
  elif compound_score <= -0.05:
    return 'Negative'
  # Scores between -0.05 and 0.05 are classified as Neutral.
  else:
    return 'Neutral'


# Map the helper function across all calculated compound scores.
df['vader_sentiment'] = df['compound'].apply(classify_sentiment)


# ==============================================================================
# STEP 4: CATEGORIZE SPECIFIC BUSINESS THEMES (Keyword Extraction)
# ==============================================================================
# Rule-based business logic function to bucket negative feedback into actionable product categories.
def map_business_theme(row):
  text = row['review_text'].lower()
  rating = row['stars_numeric']
  sentiment = row['vader_sentiment']

  # Target reviews that indicate customer friction (1 or 2 stars, or negative sentiment score).
  if rating <= 2 or sentiment == 'Negative':
    if any(
        kw in text
        for kw in [
            'estimate',
            'quote',
            'misleading',
            'code',
            'flow',
            'agree',
            'inquiry',
        ]
    ):
      return 'Onboarding / Quote Misunderstanding'
    elif any(
        kw in text
        for kw in [
            'auto',
            'renew',
            'recurring',
            'year after year',
            'rolled',
            'continue',
        ]
    ):
      return 'Unexpected Auto-Renewal'
    elif any(
        kw in text
        for kw in [
            'homestead',
            'exemption',
            'cap',
            'wrong amount',
            'exaggerate',
            'overcharge',
        ]
    ):
      return 'Exemption & Fee Disputes'
    elif any(
        kw in text
        for kw in [
            'contact',
            'phone',
            'support',
            'reach',
            'response',
            'call',
            'rep',
        ]
    ):
      return 'Customer Support Reachability'
    else:
      return 'General Dissatisfaction'
  else:
    return 'Positive / Successful Savings'


# Apply theme mapping across each row of the dataset.
df['business_theme'] = df.apply(map_business_theme, axis=1)


# ==============================================================================
# STEP 5: VISUALIZE RESULTS WITH INTERACTIVE PLOTLY CHARTS
# ==============================================================================
# Chart 1: Box Plot showing distribution of Compound Sentiment vs. Numeric Star Ratings
fig1 = px.box(
    df,
    x='stars_numeric',
    y='compound',
    color='stars_numeric',
    title='<b>VADER Compound Sentiment Score vs. Star Rating</b>',
    labels={
        'stars_numeric': 'Star Rating',
        'compound': 'Compound Sentiment Score',
    },
    color_discrete_sequence=px.colors.qualitative.Set2,
)
fig1.update_layout(template='plotly_white')
fig1.show()

# Chart 2: Horizontal Bar Chart highlighting the root causes of negative customer friction
neg_df = df[df['business_theme'] != 'Positive / Successful Savings']
theme_counts = neg_df['business_theme'].value_counts().reset_index()
theme_counts.columns = ['business_theme', 'count']

fig2 = px.bar(
    theme_counts,
    x='count',
    y='business_theme',
    orientation='h',
    title='<b>Root Cause Categories of Customer Friction</b>',
    labels={'count': 'Number of Reviews', 'business_theme': 'Friction Category'},
    color='count',
    color_continuous_scale='Reds',
)
fig2.update_layout(template='plotly_white', showlegend=False)
fig2.show()


# ==============================================================================
# STEP 6: EXPORT AGGREGATED METRICS SUMMARY
# ==============================================================================
# Create a pivot table summarizing volume, average rating, and compound sentiment by business theme.
summary_metrics = (
    df.groupby('business_theme')
    .agg(
        review_count=('review_id', 'count'),
        avg_star_rating=('stars_numeric', 'mean'),
        avg_sentiment_compound=('compound', 'mean'),
    )
    .reset_index()
)

# Format numerical floating point values to 2 decimal places for clean display.
summary_metrics['avg_star_rating'] = summary_metrics['avg_star_rating'].round(2)
summary_metrics['avg_sentiment_compound'] = summary_metrics[
    'avg_sentiment_compound'
].round(2)

# Save summary to a CSV file.
summary_metrics.to_csv('ownwell_sentiment_summary.csv', index=False)
print('Execution Complete! Summary metrics saved to "ownwell_sentiment_summary.csv"')
```

```python
# ==============================================================================
# BAG OF WORDS (BoW) ANALYSIS ON NEGATIVE REVIEWS (1 & 2 STARS)
# ==============================================================================
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS, CountVectorizer

# 1. Filter negative reviews
bad_reviews = df[df['stars_numeric'] <= 2]['review_text'].dropna()

# 2. Define custom stop words to filter out generic domain words & company names
domain_stopwords = [
    'ownwell',
    'tax',
    'taxes',
    'property',
    'service',
    'company',
    'year',
    'did',
    'don',
    'just',
    'like',
    've',
    'got',
    'said',
    'tell',
    'make',
    'know',
    'way',
    'com',
    '512',
    '886',
    '2282',
    '100',
    '2023',
    '2024',
    '2022',
    'stars',
    'star',
]
custom_stopwords = list(ENGLISH_STOP_WORDS.union(domain_stopwords))

# 3. Initialize CountVectorizers for Unigrams (single words) & Bigrams (2-word phrases)
vectorizer_uni = CountVectorizer(stop_words=custom_stopwords, min_df=2)
vectorizer_bi = CountVectorizer(
    stop_words=custom_stopwords, ngram_range=(2, 2), min_df=2
)

# 4. Transform negative review text into Bag of Words sparse matrices
X_uni = vectorizer_uni.fit_transform(bad_reviews)
X_bi = vectorizer_bi.fit_transform(bad_reviews)

# 5. Sum word occurrences across all bad reviews
top_unigrams = (
    pd.DataFrame(X_uni.toarray(), columns=vectorizer_uni.get_feature_names_out())
    .sum()
    .sort_values(ascending=False)
    .head(15)
)

top_bigrams = (
    pd.DataFrame(X_bi.toarray(), columns=vectorizer_bi.get_feature_names_out())
    .sum()
    .sort_values(ascending=False)
    .head(15)
)

# 6. Plot the Top Unigrams and Bigrams
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

top_unigrams.sort_values().plot(
    kind='barh', ax=axes[0], color='#d9534f', edgecolor='black'
)
axes[0].set_title(
    '<b>Top 15 Complaint Keywords (Unigrams)</b>', fontsize=12
)
axes[0].set_xlabel('Word Count Frequency')

top_bigrams.sort_values().plot(
    kind='barh', ax=axes[1], color='#337ab7', edgecolor='black'
)
axes[1].set_title(
    '<b>Top 15 Complaint Phrases (Bigrams)</b>', fontsize=12
)
axes[1].set_xlabel('Phrase Count Frequency')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# 1. Load pre-trained Sentence Transformer model
model = SentenceTransformer('all-MiniLM-L6-v2')

# 2. Define specific recurring claim hypotheses to test against negative reviews
candidate_claims = [
    'User thought they were getting a quote but received a binding invoice',
    'Billed fee on automatic homestead exemption without adding value',
    'Unaware that agreement auto-renews year after year',
    'Customer support is unresponsive or unreachable by phone/email',
]

# 3. Filter bad reviews
bad_reviews = df[df['stars_numeric'] <= 2]['review_text'].tolist()

# 4. Generate embeddings
review_embeddings = model.encode(bad_reviews)
claim_embeddings = model.encode(candidate_claims)

# 5. Compute cosine similarities (Matrix multiplication)
similarity_matrix = cosine_similarity(review_embeddings, claim_embeddings)

# 6. Assign the highest-scoring claim hypothesis to each review
matched_claims = []
for idx, review in enumerate(bad_reviews):
  top_claim_idx = np.argmax(similarity_matrix[idx])
  confidence = similarity_matrix[idx][top_claim_idx]
  matched_claims.append({
      'review_text': review[:100] + '...',
      'predicted_claim': candidate_claims[top_claim_idx],
      'confidence_score': round(float(confidence), 3),
  })

results_df = pd.DataFrame(matched_claims)
print(results_df.head())
```

```python
# ==============================================================================
# EXACT CLAIM EXTRACTION VIA OPENROUTER API
# ==============================================================================
import os
import json
import pandas as pd
from google.colab import userdata
from openai import OpenAI

# 1. Retrieve key from Colab secrets or environment variable
try:
  api_key = userdata.get('openrouter_api_key')
except Exception:
  api_key = os.getenv('openrouter_api_key')

# 2. Initialize OpenAI client configured for OpenRouter endpoint
client = OpenAI(
    base_url='https://openrouter.ai/api/v1',
    api_key=api_key,
)

# 3. Filter negative reviews directly from notebook's `df`
bad_reviews_df = df[df['stars_numeric'] <= 2].copy()
bad_reviews_df['review_text'] = bad_reviews_df['review_text'].fillna('')


# 4. Function to classify claims and risk level using OpenRouter
def extract_claim_openrouter(review_text):
  system_prompt = """
    You are an expert NLP product analyst. Analyze the following negative review and extract:
    1. "exact_claim": Select EXACTLY ONE of these categories:
       - "quote_to_invoice": User expected a free estimate/quote but received a binding bill/invoice.
       - "auto_renewal_surprise": User was unaware of annual recurring auto-renewal.
       - "homestead_exemption_dispute": User was charged a fee on automatic homestead exemption savings.
       - "unresponsive_support": User could not reach support or get an email response.
       - "dissatisfied_savings": Dissatisfied with the tax reduction amount or property appraisal result.
    2. "churn_risk_score": Integer from 1 to 3 (1 = Low, 2 = Moderate, 3 = High Churn/Fraud/Chargeback Risk).

    Return strictly a JSON object with keys "exact_claim" and "churn_risk_score".
    """

  try:
    response = client.chat.completions.create(
        model='openai/gpt-4o-mini',  # Fast, highly accurate, and extremely cheap via OpenRouter
        response_format={'type': 'json_object'},
        temperature=0.0,
        messages=[
            {'role': 'system', 'content': system_prompt},
            {'role': 'user', 'content': f'Review: "{review_text}"'},
        ],
    )
    res_json = json.loads(response.choices[0].message.content)
    return res_json.get('exact_claim', 'Unclassified'), res_json.get(
        'churn_risk_score', 1
    )
  except Exception as e:
    return f'API Error: {str(e)[:30]}', 1


# 5. Process sample batch of bad reviews (remove .head(10) to run across all bad reviews)
print('Running OpenRouter Claim Extraction Engine...')

claims = []
risk_scores = []

# Using head(10) for demonstration speed
sample_df = bad_reviews_df.head(10).copy()

for idx, row in sample_df.iterrows():
  claim, risk = extract_claim_openrouter(row['review_text'])
  claims.append(claim)
  risk_scores.append(risk)

sample_df['openrouter_claim'] = claims
sample_df['churn_risk_score'] = risk_scores

# 6. Display Structured Output
print('\n=== OPENROUTER EXTRACTION RESULTS ===')
print(
    sample_df[[
        'stars_numeric',
        'openrouter_claim',
        'churn_risk_score',
        'review_text',
    ]].to_string()
)
```

```python
# ==============================================================================
# 1. RUN OPENROUTER ENGINE ACROSS ALL BAD REVIEWS (200+ REVIEWS)
# ==============================================================================
print('Running OpenRouter extraction across all negative reviews...')

claims = []
risk_scores = []

# Process all 1-star and 2-star reviews
for idx, row in bad_reviews_df.iterrows():
  claim, risk = extract_claim_openrouter(row['review_text'])
  claims.append(claim)
  risk_scores.append(risk)

bad_reviews_df['openrouter_claim'] = claims
bad_reviews_df['churn_risk_score'] = risk_scores

# Merge findings back into main `df` DataFrame
df['openrouter_claim'] = 'Positive / Satisfied'
df['churn_risk_score'] = 1
df.loc[bad_reviews_df.index, 'openrouter_claim'] = bad_reviews_df[
    'openrouter_claim'
]
df.loc[bad_reviews_df.index, 'churn_risk_score'] = bad_reviews_df[
    'churn_risk_score'
]

# ==============================================================================
# 2. AGGREGATE & VISUALIZE THE OPENROUTER CLAIMS
# ==============================================================================
import plotly.express as px

# Aggregated summary table
summary_df = (
    bad_reviews_df.groupby('openrouter_claim')
    .agg(
        review_count=('review_id', 'count'),
        avg_churn_risk=('churn_risk_score', 'mean'),
    )
    .reset_index()
    .sort_values(by='review_count', ascending=False)
)

print('\n=== AGGREGATED OPENROUTER CLAIM BREAKDOWN ===')
print(summary_df)

# Plot interactive horizontal bar chart
fig = px.bar(
    summary_df,
    x='review_count',
    y='openrouter_claim',
    color='avg_churn_risk',
    orientation='h',
    title=(
        '<b>OpenRouter Extracted Claims vs. Average Churn/Fraud Risk'
        ' Score</b>'
    ),
    labels={
        'review_count': 'Number of Negative Reviews',
        'openrouter_claim': 'Extracted Specific Claim',
        'avg_churn_risk': 'Avg Churn Risk (1-3)',
    },
    color_continuous_scale='Reds',
)
fig.update_layout(template='plotly_white')
fig.show()

# Export finalized dataset
df.to_csv('ownwell_full_openrouter_analysis.csv', index=False)
print(
    '\nFinal dataset saved successfully to "ownwell_full_openrouter_analysis.csv"'
)
```

# Executive Synthesis & Case Study: Ownwell Customer Sentiment Analysis

## 1. Executive Summary
An NLP-driven sentiment analysis of **3,090 customer reviews** revealed a overall high customer satisfaction score (**89.4% 5-star reviews**). However, a detailed analysis of all low-rating feedback (**218 negative reviews**, representing 7.1% of total volume) identified critical onboarding, billing, and retention friction points.

Using OpenRouter and `gpt-4o-mini` with strict structured JSON schema outputs, each negative review was classified into exact operational claims and assigned a Churn/Legal Risk Score (1.0 = Low, 3.0 = High).

## 3. Key Findings & Strategic Insights

### A. The Core Friction Engine: Savings Disappointment + Support Bottlenecks
* **Volume Leader (77.5% combined):** `dissatisfied_savings` (94 reviews) and `unresponsive_support` (75 reviews) account for the overwhelming majority of low-star ratings.
* **The Escalation Path:** Dissatisfaction with tax protest outcomes often morphs into public 1-star reviews only after customers hit support wall (unresponsive email/phone channels or automated loops).

### B. Highest Churn & Brand Risk: Onboarding & Billing Transparency
* **Severe Churn/Fraud Indicators (2.95 - 2.96 Risk Scores):** While lower in volume, `auto_renewal_surprise` (24 reviews) and `quote_to_invoice` (19 reviews) carry the highest risk of chargebacks, Better Business Bureau (BBB) complaints, and accusations of predatory billing.
* **Root Cause:** Users entering the top-of-funnel expecting a simple tax estimate are entering binding agreements without explicit, multi-step confirmation.
